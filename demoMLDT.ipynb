```python
def calculateBoundaryConstant_A(n_tissue, n_outside=1.0):
    """
    Calculates the exact partial-current boundary constant A by integrating 
    unpolarized Fresnel reflection coefficients over the hemisphere.
    """
    # Snell's Law and Fresnel reflection for light exiting the tissue
    def fresnel_unpol(theta_i):
        # Check for total internal reflection
        if np.sin(theta_i) >= (n_outside / n_tissue):
            return 1.0
            
        # Calculate transmission angle
        theta_t = np.arcsin((n_tissue / n_outside) * np.sin(theta_i))
        
        # S-polarized and P-polarized reflection coefficients
        Rs = ((n_tissue * np.cos(theta_i) - n_outside * np.cos(theta_t)) / 
              (n_tissue * np.cos(theta_i) + n_outside * np.cos(theta_t)))**2
              
        Rp = ((n_outside * np.cos(theta_i) - n_tissue * np.cos(theta_t)) / 
              (n_outside * np.cos(theta_i) + n_tissue * np.cos(theta_t)))**2
        return 0.5 * (Rs + Rp)

    # First angular moment (R_phi)
    def int_R_phi(theta):
        return 2.0 * np.sin(theta) * np.cos(theta) * fresnel_unpol(theta)
        
    # Second angular moment (R_j)
    def int_R_j(theta):
        return 3.0 * np.sin(theta) * (np.cos(theta)**2) * fresnel_unpol(theta)

    # Integrate over the hemisphere (0 to pi/2)
    R_phi, _ = quad(int_R_phi, 0, np.pi/2)
    R_j, _ = quad(int_R_j, 0, np.pi/2)
    
    # Calculate final proportionality constant A
    A = (1.0 - R_phi) / (2.0 * (1.0 + R_j))
    
    return A

print(calculateBoundaryConstant_A(n_tissue=1.4, n_outside=1.0))
```

```python
def getRd_MLDT(mua_layers, musp_layers, thicknesses, nr):
    ###################################################
    #### 1. SETUP
    ###################################################
    eps = 1e-10
    mua_layers = np.clip(np.asarray(mua_layers, dtype=float), eps, None)
    musp_layers = np.clip(np.asarray(musp_layers, dtype=float), eps, None)
    thicknesses = np.clip(np.asarray(thicknesses, dtype=float), eps, None)
    nr = float(nr)

    # The function was not set-up to handle 1-layer models.
    # This statement creates a 2-layer model with identical 
    # optical properties if only 1 layer was used as input.
    if thicknesses.size == 0:
        # Give the first layer an arbitrary thickness
        # Since properties are identical, this interface is transparent.
        thicknesses = np.array([10.0]) 
        
        # Duplicate the optical properties for the second layer
        mua_layers = np.column_stack((mua_layers, mua_layers))
        musp_layers = np.column_stack((musp_layers, musp_layers))
    
    Nnm, num_total_layers = mua_layers.shape
    num_finite_layers = len(thicknesses)
    num_coeffs = 2 * num_finite_layers + 1
    
    ###################################################
    #### 2. CALCULATE COEFFICIENTS
    ###################################################
    mutr_all = mua_layers + musp_layers
    zeta_all = 1.0 / (3.0 * mutr_all)

    denom_delta = 3 * mua_layers * mutr_all
    delta_all = 1 / np.sqrt(np.clip(denom_delta, eps, None))
    
    A = calculateBoundaryConstant_A(nr, n_outside=1.0)
    Phi_t = 1.0

    # Source term S0
    res_denom = 1 - (mutr_all**2 * delta_all**2)
    res_denom = np.where(np.abs(res_denom) < eps, eps, res_denom) # avoid 0

    S0_base_all = ((Phi_t * delta_all**2 * musp_layers) /
                   (zeta_all * res_denom))

    ###################################################
    #### 3. ATTENUATION
    ###################################################
    mutr_prev = mutr_all[:, :-1] 
    thick_prev = thicknesses     
    att_factors = np.exp(-mutr_prev * thick_prev[None, :])
    ones_col = np.ones((Nnm, 1))
    cumulative_att = np.hstack((ones_col, np.cumprod(att_factors, axis=1)))
    S0_attenuated_all = S0_base_all * cumulative_att

    ###################################################
    #### 4. PREPARE BANDED MATRIX
    ###################################################
    # ab shape: (Nnm, 5, num_coeffs)
    # Row 0: Top (j=i+2), Row 1: Upper (j=i+1), Row 2: Main (j=i), Row 3: Lower (j=i-1), Row 4: Bottom (j=i-2)
    ab = np.zeros((Nnm, 5, num_coeffs), dtype=float)
    B_vec = np.zeros((Nnm, num_coeffs), dtype=float)
    
    k = np.arange(num_finite_layers)
    
    ###################################################
    #### 5. FILL DIAGONALS
    ###################################################
    #### 5a. Surface Boundary (Row 0)
    # A[0,0] -> Main (Row 2, Col 0)
    # A[0,1] -> Upper (Row 1, Col 1)
    ab[:, 2, 0] = (zeta_all[:,0] / delta_all[:,0]) + A
    ab[:, 1, 1] = (-zeta_all[:,0] / delta_all[:,0]) + A
    B_vec[:, 0] = -(A * S0_attenuated_all[:,0]) - (zeta_all[:,0] * S0_attenuated_all[:,0] * mutr_all[:,0])

    #### 5b. Internal Interfaces
    zeta_k = zeta_all[:, k]; delta_k = delta_all[:, k]
    mutr_k = mutr_all[:, k]; S0_att_k = S0_attenuated_all[:, k]
    zeta_kp1 = zeta_all[:, k+1]; delta_kp1 = delta_all[:, k+1]
    mutr_kp1 = mutr_all[:, k+1]; S0_att_kp1 = S0_attenuated_all[:, k+1]
    
    dz_k = thicknesses[None, :] 
    exp_dec_k = np.exp(-dz_k / delta_k)
    exp_gro_k = np.exp(dz_k / delta_k)
    
    src_k = S0_att_k * np.exp(-mutr_k * dz_k)
    src_kp1 = S0_att_kp1
    d_src_k = -S0_att_k * mutr_k * np.exp(-mutr_k * dz_k)
    d_src_kp1 = -S0_att_kp1 * mutr_kp1
    
    # Column Indices
    col_k_dec = 2 * k
    col_k_gro = 2 * k + 1
    col_kp1_dec = 2 * (k + 1)
    col_kp1_gro = 2 * (k + 1) + 1
    
    #### Equation 2k+1 (Fluence Continuity)
    # C_k_dec (Col 2k): j = i-1 -> Lower (Row 3)
    ab[:, 3, col_k_dec] = exp_dec_k 
    
    # C_k_gro (Col 2k+1): j = i -> Main (Row 2)
    ab[:, 2, col_k_gro] = exp_gro_k
    
    # C_kp1_dec (Col 2k+2): j = i+1 -> Upper (Row 1)
    ab[:, 1, col_kp1_dec] = -1.0
    
    # C_kp1_gro (Col 2k+3): j = i+2 -> Top (Row 0)
    mask_fin = k < (num_finite_layers - 1)
    ab[:, 0, col_kp1_gro[mask_fin]] = -1.0
    
    B_vec[:, 2*k+1] = src_kp1 - src_k

    #### Equation 2k+2 (Flux Continuity)
    term_k_dec = -zeta_k * (-exp_dec_k / delta_k)
    term_k_gro = -zeta_k * (exp_gro_k / delta_k)
    term_kp1_dec = zeta_kp1 * (-1.0 / delta_kp1)
    term_kp1_gro = zeta_kp1 * (1.0 / delta_kp1)

    # C_k_dec (Col 2k): j = i-2 -> Bottom (Row 4)
    ab[:, 4, col_k_dec] = term_k_dec
    
    # C_k_gro (Col 2k+1): j = i-1 -> Lower (Row 3)
    ab[:, 3, col_k_gro] = term_k_gro
    
    # C_kp1_dec (Col 2k+2): j = i -> Main (Row 2)
    ab[:, 2, col_kp1_dec] = term_kp1_dec
    
    # C_kp1_gro (Col 2k+3): j = i+1 -> Upper (Row 1)
    ab[:, 1, col_kp1_gro[mask_fin]] = term_kp1_gro[:, mask_fin]
    
    B_vec[:, 2*k+2] = -zeta_kp1 * d_src_kp1 + zeta_k * d_src_k

    ###################################################
    #### 6. SOLVE
    ###################################################
    C_coeffs = np.zeros((Nnm, num_coeffs))
    
    # Loop over wavelengths
    for i in range(Nnm):
        C_coeffs[i, :] = solve_banded((2, 2), ab[i], B_vec[i], overwrite_ab=True)
    
    ###################################################
    #### 7. REFLECTANCE
    ###################################################
    phi_0 = S0_attenuated_all[:, 0] + C_coeffs[:, 0] + C_coeffs[:, 1]
    j_0 = -A * phi_0
    Rd = -(j_0 / Phi_t)
    
    return Rd
```

## N-layer Depth Variation

```python
def create_subrange(min_val, max_val, instance_vals=[], set_instance_min=False, set_instance_max=False):
    if set_instance_min == False and set_instance_max == False:
        r1 = np.random.uniform(min_val, max_val)
        r2 = np.random.uniform(min_val, max_val)
        instance_min = min(r1, r2)
        instance_max = max(r1, r2)
    elif set_instance_min == True and set_instance_max == False:
        instance_min = instance_vals[0]
        instance_max = np.random.uniform(instance_min, max_val)
    elif set_instance_min == False and set_instance_max == True:
        instance_max = instance_vals[1]
        instance_min = np.random.uniform(min_val, instance_max)
    elif set_instance_min == True and set_instance_max == True:
        instance_min = instance_vals[0]
        instance_max = instance_vals[1]

    min_spread = (max_val - min_val) * 0.05
    if (instance_max - instance_min) < min_spread:
        center = (instance_max + instance_min) / 2
        instance_min = max(min_val, center - min_spread/2)
        instance_max = min(max_val, center + min_spread/2)

    return instance_min, instance_max

def generate_gaussian_profile(num_points, instance_min, instance_max, min_peaks, max_peaks, center_min, center_max, std_dev_min, std_dev_max):
    """
    Generates a randomized Gaussian-based profile that floats within a random sub-range 
    bounded by [instance_min, instance_max].
    """
    profile = np.zeros(num_points)
    num_peaks = np.random.randint(min_peaks, max_peaks + 1)
    
    for _ in range(num_peaks):
        center = np.random.uniform(center_min, center_max)
        amplitude = np.random.uniform(0.1, 1.0) 
        std_dev = np.random.uniform(std_dev_min * num_points, std_dev_max * num_points)

        x = np.arange(num_points)
        gaussian = amplitude * np.exp(-((x - center)**2) / (2 * std_dev**2))
        profile += gaussian

    # --- Normalize the Shape ---
    profile = (profile - np.min(profile)) / (np.max(profile) - np.min(profile))
    
    # --- Apply the Range ---
    # Scale: (0 to 1) -> (instance_min to instance_max)
    profile = profile * (instance_max - instance_min) + instance_min
    
    return profile
```
